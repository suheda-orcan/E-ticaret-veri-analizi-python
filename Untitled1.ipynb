```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Görselleştirme ayarları
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Örnek e-ticaret veri setini GitHub üzerinden yüklüyoruz
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
df = pd.read_csv(url, sep='\t')

# Verinin ilk 5 satırına göz atalım
print("Veri başarıyla yüklendi! İşte ilk 5 satır:")
df.head()
```

```python
print(f"Toplam Satır Sayısı: {df.shape[0]}")
print(f"Toplam Sütun Sayısı: {df.shape[1]}\n")

print("--- Sütun Bilgileri ve Veri Tipleri ---")
df.info()
```

```python

# 1. 'item_price' sütunundaki '$' işaretini temizleyip float (ondalıklı sayı) yapıyoruz
df['item_price'] = df['item_price'].str.replace('$', '', regex=False).astype(float)

# 2. Her ürün için harcanan Toplam Tutarı (Quantity * Item Price) hesaplayıp yeni bir sütun ekleyelim
df['total_price'] = df['quantity'] * df['item_price']

# 3. Eksik (null) değer kontrolü yapalım
print("--- Sütunlardaki Eksik Değer Sayıları ---")
print(df.isnull().sum())

# 4. Değişiklikleri doğrulamak için veri tiplerini ve ilk 5 satırı tekrar görelim
print("\n--- Güncellenmiş Veri Tipleri ---")
print(df.dtypes[['item_price', 'total_price']])

df.head()
```

```python
# 1. En çok sipariş edilen ilk 5 ürün hangisi?
top_products = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(5)
print("--- En Çok Satan İlk 5 Ürün ---")
print(top_products)

# 2. Toplam kaç farklı sipariş verilmiş ve toplam ne kadar ciro elde edilmiş?
total_orders = df['order_id'].nunique()
total_revenue = df['total_price'].sum()
avg_order_value = total_revenue / total_orders

print("\n--- Özet Finansal Metrikler ---")
print(f"Toplam Sipariş Sayısı: {total_orders}")
print(f"Toplam Ciro: ${total_revenue:,.2f}")
print(f"Sipariş Başına Ortalama Harcama (AOV): ${avg_order_value:.2f}")
```

```python
# Yan yana iki grafik çizdirmek için figür oluşturuyoruz
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Grafik: En Çok Satan İlk 5 Ürün (Adet Bazında)
top_qty = df.groupby('item_name')['quantity'].sum().sort_values(ascending=False).head(5)
sns.barplot(x=top_qty.values, y=top_qty.index, ax=axes[0], palette="Blues_r")
axes[0].set_title("En Çok Satılan İlk 5 Ürün (Adet)", fontsize=14, fontweight='bold')
axes[0].set_xlabel("Toplam Satış Adedi")
axes[0].set_ylabel("Ürün Adı")

# 2. Grafik: En Çok Ciro Getiren İlk 5 Ürün (Dolar Bazında)
top_rev = df.groupby('item_name')['total_price'].sum().sort_values(ascending=False).head(5)
sns.barplot(x=top_rev.values, y=top_rev.index, ax=axes[1], palette="Greens_r")
axes[1].set_title("En Çok Ciro Getiren İlk 5 Ürün ($)", fontsize=14, fontweight='bold')
axes[1].set_xlabel("Toplam Ciro ($)")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()
```
